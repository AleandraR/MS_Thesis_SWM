# 3. LSWMI

```python
# Load everything
wsi  = xr.open_dataset("C:/Users/Nitro 5/Documents/MS/Thesis/GitHub/MS_Thesis_SWM/01_data/02_processed/LSWMI_Prep-WSI.nc")
ssi  = xr.open_dataset("C:/Users/Nitro 5/Documents/MS/Thesis/GitHub/MS_Thesis_SWM/01_data/02_processed/LSWMI_Prep-SSI.nc")
msl = xr.open_dataset("C:/Users/Nitro 5/Documents/MS/Thesis/GitHub/MS_Thesis_SWM/01_data/02_processed/LSWMI_Prep-MSLAI.nc")
ttr  = xr.open_dataset("C:/Users/Nitro 5/Documents/MS/Thesis/GitHub/MS_Thesis_SWM/01_data/02_processed/LSWMI_Prep-TTRAI.nc")
tp   = xr.open_dataset("C:/Users/Nitro 5/Documents/MS/Thesis/GitHub/MS_Thesis_SWM/01_data/02_processed/LSWMI_Prep-TPAI.nc")
```

```python
assert wsi.pentad.equals(ssi.pentad)
assert wsi.pentad.equals(msl.pentad)
assert wsi.pentad.equals(ttr.pentad)
assert wsi.pentad.equals(tp.pentad)

assert wsi.latitude.equals(ssi.latitude)
assert wsi.longitude.equals(ssi.longitude)
```

```python
wsi_u = wsi["u"].stack(time=("year", "pentad"))
ssi_v = ssi["v"].stack(time=("year", "pentad"))
msl_m = msl["msl"].stack(time=("year", "pentad"))
ttr_t = ttr["ttr"].stack(time=("year", "pentad"))
tp_r  = tp["tp"].stack(time=("year", "pentad"))

X_wsi = wsi_u.transpose("time", "latitude", "longitude").values
X_ssi = ssi_v.transpose("time", "latitude", "longitude").values
X_msl = msl_m.transpose("time", "latitude", "longitude").values
X_ttr = ttr_t.transpose("time", "latitude", "longitude").values
Y_tp  = tp_r.transpose("time", "latitude", "longitude").values
```

```python
import numpy as np
import xarray as xr
import statsmodels.api as sm

nt, ny, nx = X_wsi.shape

lswmi = np.full((nt, ny, nx), np.nan)

# Regression coefficients: [Intercept, WSI, SSI, TTR, MSL]
coeff = np.full((ny, nx, 5), np.nan)

for j in range(ny):
    for i in range(nx):

        # 1. X = coefficients sum
        X = np.column_stack([
            X_wsi[:, j, i],
            X_ssi[:, j, i],
            X_ttr[:, j, i],
            X_msl[:, j, i],
        ])

        # Y = rainfall
        y = Y_tp[:, j, i]

        if np.any(np.isnan(X)) or np.any(np.isnan(y)):
            continue

        # 2. Add intercepts
        Xc = sm.add_constant(X)

        try:
            # 3. MLR; MATLAB bisquare = Python Tukey's biweight
            rlm = sm.RLM(
                y,
                Xc,
                M=sm.robust.norms.TukeyBiweight()
            )

            res = rlm.fit()

            # 4. Store coefficients
            coeff[j, i, :] = res.params

            # 5. LSWMI
            lswmi[:, j, i] = np.dot(Xc, res.params)

        except Exception:
            continue
```

```python
# 6. Package into dataset
import os as os

lswmi_ds = xr.Dataset(
    data_vars=dict(
        # LSWMI index (linear predictor)
        lswmi=(("time", "latitude", "longitude"), lswmi),

        # Regression coefficients
        coeff=(("latitude", "longitude", "param"), coeff)
    ),
    coords=dict(
        time=wsi_u.time,

        latitude=wsi_u.latitude,
        longitude=wsi_u.longitude,

        # Regression parameters
        param=["Intercept", "WSI", "SSI", "TTR", "MSL"]
    ),
    attrs=dict(
        description="Localized Southwest Monsoon Index (LSWMI)",
        base_period="1991–2020",
        season="May–September (ESW)",
        calendar="pentad",
        method="Grid-point robust multiple linear regression",
        regression="Tukey biweight (bisquare)"
    )
)

# 7. NetCDF fix
lswmi_ds = lswmi_ds.reset_index("time")

# Pentad labels
lswmi_ds = lswmi_ds.assign_coords(
    pentad_label=(
        "time",
        [f"P{p:02d}" for p in lswmi_ds["pentad"].values]
    )
)

# Final output
lswmi_ds.to_netcdf("LSWMI_1991_2020_ESW.nc")
out_dir = r"C:/Users/Nitro 5/Documents/MS/Thesis/GitHub/MS_Thesis_SWM/01_data/02_processed/"
os.makedirs(out_dir, exist_ok=True)

out_path = os.path.join(out_dir, "LSWMI_1991_2020_ESW.nc")
lswmi_ds.to_netcdf(out_path)
```

# 6f. D_LSWMI (1991-2020 Climatology)

```python
import numpy as np
import xarray as xr
import statsmodels.api as sm
import os
```

```python
# Load DAILY predictors and rainfall
wsi  = xr.open_dataset("C:/Users/Nitro 5/Documents/MS/Thesis/GitHub/MS_Thesis_SWM/01_data/02_processed/LSWMI_Prep-D_WSI.nc")
ssi  = xr.open_dataset("C:/Users/Nitro 5/Documents/MS/Thesis/GitHub/MS_Thesis_SWM/01_data/02_processed/LSWMI_Prep-D_SSI.nc")
msl  = xr.open_dataset("C:/Users/Nitro 5/Documents/MS/Thesis/GitHub/MS_Thesis_SWM/01_data/02_processed/LSWMI_Prep-D_MSLAI.nc")
ttr  = xr.open_dataset("C:/Users/Nitro 5/Documents/MS/Thesis/GitHub/MS_Thesis_SWM/01_data/02_processed/LSWMI_Prep-D_TTRAI.nc")
tp   = xr.open_dataset("C:/Users/Nitro 5/Documents/MS/Thesis/GitHub/MS_Thesis_SWM/01_data/02_processed/LSWMI_Prep-D_TPAI.nc")
```

```python
assert wsi.valid_time.equals(ssi.valid_time)
assert wsi.valid_time.equals(msl.valid_time)
assert wsi.valid_time.equals(ttr.valid_time)
assert wsi.valid_time.equals(tp.valid_time)

assert wsi.latitude.equals(ssi.latitude)
assert wsi.longitude.equals(ssi.longitude)
```

```python
X_wsi = wsi["WSI"].transpose("valid_time", "latitude", "longitude").values
X_ssi = ssi["SSI"].transpose("valid_time", "latitude", "longitude").values
X_msl = msl["MSLAI"].transpose("valid_time", "latitude", "longitude").values
X_ttr = ttr["TTRAI"].transpose("valid_time", "latitude", "longitude").values
Y_tp  = tp["TPAI"].transpose("valid_time", "latitude", "longitude").values
```

```python
times = wsi.valid_time.values
mask_base = (times >= np.datetime64('1991-05-01')) & (times <= np.datetime64('2020-09-30'))

nt, ny, nx = X_wsi.shape

lswmi = np.full((nt, ny, nx), np.nan)
coeff = np.full((ny, nx, 5), np.nan)  # Intercept + 4 predictors

for j in range(ny):
    for i in range(nx):

        # 1️⃣ Stack predictors for this grid point
        X_full = np.column_stack([
            X_wsi[:, j, i],
            X_ssi[:, j, i],
            X_ttr[:, j, i],
            X_msl[:, j, i],
        ])
        y_full = Y_tp[:, j, i]

        X_train = X_full[mask_base, :]
        y_train = y_full[mask_base]

        # 2️⃣ Skip if any NaNs
        if np.any(np.isnan(X_train)) or np.any(np.isnan(y_train)):
            continue

        # 3️⃣ Skip if any predictor has near-zero variance
        if np.any(np.nanstd(X_train, axis=0) < 1e-6):
            continue

        # 4️⃣ Standardize predictors (mean=0, std=1)
        mu_base  = np.mean(X_train, axis=0)
        std_base = np.std(X_train, axis=0, ddof=1)

        X_full_std = (X_full - mu_base) / std_base

        # 5️⃣ Add constant (intercept)
        Xc_full = sm.add_constant(X_full_std)

        Xc_train = Xc_full[mask_base, :]

        try:
            rlm = sm.RLM(
                y_train,
                Xc_train,
                M=sm.robust.norms.TukeyBiweight()
            )
            res = rlm.fit()

            lswmi[:, j, i] = Xc_full @ res.params

        except Exception:
            continue
```

```python
lswmi_ds = xr.Dataset(
    data_vars=dict(
        lswmi=(("valid_time", "latitude", "longitude"), lswmi),
        coeff=(("latitude", "longitude", "param"), coeff),
    ),
    coords=dict(
        valid_time=wsi.valid_time, # Contains full 1979-2025
        latitude=wsi.latitude,
        longitude=wsi.longitude,
        param=["Intercept", "WSI", "SSI", "TTR", "MSL"]
    ),
    attrs=dict(
        description="Daily Local Southwest Monsoon Index (D-LSWMI)",
        period="1979–2025",
        base_period="1991–2020",
        season="May–September (ESW)",
        method="Grid-point robust regression (Trained on 1991-2020, Applied to 1979-2025)",
        regression="Tukey biweight (bisquare)"
    )
)

base_dir = r"C:/Users/Nitro 5/Documents/MS/Thesis/GitHub/MS_Thesis_SWM/01_data/02_processed/"
os.makedirs(base_dir, exist_ok=True)

out_path = os.path.join(base_dir, "D_LSWMI_1979_2025_ESW.nc")
lswmi_ds.to_netcdf(out_path)
print(f"Saved: {out_path}")
```
