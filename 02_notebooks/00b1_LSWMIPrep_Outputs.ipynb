# 3. LSWMI

```python
# Load everything
wsi  = xr.open_dataset("C:/Users/Nitro 5/Documents/MS/Thesis/GitHub/MS_Thesis_SWM/01_data/02_processed/LSWMI_Prep-WSI.nc")
ssi  = xr.open_dataset("C:/Users/Nitro 5/Documents/MS/Thesis/GitHub/MS_Thesis_SWM/01_data/02_processed/LSWMI_Prep-SSI.nc")
msl = xr.open_dataset("C:/Users/Nitro 5/Documents/MS/Thesis/GitHub/MS_Thesis_SWM/01_data/02_processed/LSWMI_Prep-MSLAI.nc")
ttr  = xr.open_dataset("C:/Users/Nitro 5/Documents/MS/Thesis/GitHub/MS_Thesis_SWM/01_data/02_processed/LSWMI_Prep-TTRAI.nc")
tp   = xr.open_dataset("C:/Users/Nitro 5/Documents/MS/Thesis/GitHub/MS_Thesis_SWM/01_data/02_processed/LSWMI_Prep-TPAI.nc")
```

```python
assert wsi.pentad.equals(ssi.pentad)
assert wsi.pentad.equals(msl.pentad)
assert wsi.pentad.equals(ttr.pentad)
assert wsi.pentad.equals(tp.pentad)

assert wsi.latitude.equals(ssi.latitude)
assert wsi.longitude.equals(ssi.longitude)
```

```python
wsi_u = wsi["u"].stack(time=("year", "pentad"))
ssi_v = ssi["v"].stack(time=("year", "pentad"))
msl_m = msl["msl"].stack(time=("year", "pentad"))
ttr_t = ttr["ttr"].stack(time=("year", "pentad"))
tp_r  = tp["tp"].stack(time=("year", "pentad"))

X_wsi = wsi_u.transpose("time", "latitude", "longitude").values
X_ssi = ssi_v.transpose("time", "latitude", "longitude").values
X_msl = msl_m.transpose("time", "latitude", "longitude").values
X_ttr = ttr_t.transpose("time", "latitude", "longitude").values
Y_tp  = tp_r.transpose("time", "latitude", "longitude").values
```

```python
import numpy as np
import xarray as xr
import statsmodels.api as sm

nt, ny, nx = X_wsi.shape

lswmi = np.full((nt, ny, nx), np.nan)

# Regression coefficients: [Intercept, WSI, SSI, TTR, MSL]
coeff = np.full((ny, nx, 5), np.nan)

for j in range(ny):
    for i in range(nx):

        # 1. X = coefficients sum
        X = np.column_stack([
            X_wsi[:, j, i],
            X_ssi[:, j, i],
            X_ttr[:, j, i],
            X_msl[:, j, i],
        ])

        # Y = rainfall
        y = Y_tp[:, j, i]

        if np.any(np.isnan(X)) or np.any(np.isnan(y)):
            continue

        # 2. Add intercepts
        Xc = sm.add_constant(X)

        try:
            # 3. MLR; MATLAB bisquare = Python Tukey's biweight
            rlm = sm.RLM(
                y,
                Xc,
                M=sm.robust.norms.TukeyBiweight()
            )

            res = rlm.fit()

            # 4. Store coefficients
            coeff[j, i, :] = res.params

            # 5. LSWMI
            lswmi[:, j, i] = np.dot(Xc, res.params)

        except Exception:
            continue
```

```python
# 6. Package into dataset
import os as os

lswmi_ds = xr.Dataset(
    data_vars=dict(
        # LSWMI index (linear predictor)
        lswmi=(("time", "latitude", "longitude"), lswmi),

        # Regression coefficients
        coeff=(("latitude", "longitude", "param"), coeff)
    ),
    coords=dict(
        time=wsi_u.time,

        latitude=wsi_u.latitude,
        longitude=wsi_u.longitude,

        # Regression parameters
        param=["Intercept", "WSI", "SSI", "TTR", "MSL"]
    ),
    attrs=dict(
        description="Localized Southwest Monsoon Index (LSWMI)",
        base_period="1991–2020",
        season="May–September (ESW)",
        calendar="pentad",
        method="Grid-point robust multiple linear regression",
        regression="Tukey biweight (bisquare)"
    )
)

# 7. NetCDF fix
lswmi_ds = lswmi_ds.reset_index("time")

# Pentad labels
lswmi_ds = lswmi_ds.assign_coords(
    pentad_label=(
        "time",
        [f"P{p:02d}" for p in lswmi_ds["pentad"].values]
    )
)

# Final output
lswmi_ds.to_netcdf("LSWMI_1991_2020_ESW.nc")
out_dir = r"C:/Users/Nitro 5/Documents/MS/Thesis/GitHub/MS_Thesis_SWM/01_data/02_processed/"
os.makedirs(out_dir, exist_ok=True)

out_path = os.path.join(out_dir, "LSWMI_1991_2020_ESW.nc")
lswmi_ds.to_netcdf(out_path)
```

```python
lswmi_da = xr.DataArray(
    lswmi,
    dims=("time", "latitude", "longitude"),
    coords={
        "time": wsi_u.time,
        "latitude": wsi.latitude,
        "longitude": wsi.longitude,
    },
    name="lswmi"
)
```
